```python
text="who is Pm of india?"
```

```python
import tiktoken
```

```python
#object creation
tokenizer=tiktoken.encoding_for_model(model_name='gpt-4')
```

```python
tokenids=tokenizer.encode(text)
tokenids
```

with the help of text ,tokenizer object was created then from tokens tokeids were calculated and passed to llm

```python
resultids=[791,393,76,315,28811,374,5768,35973,13]
# the resultids generated by llm as answer now will be converted to text which will be visible as answer on ui
```
